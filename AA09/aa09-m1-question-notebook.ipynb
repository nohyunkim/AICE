### **5. 윤활유 압력(kPa)(oilPressure) 컬럼의 극단값을 상·하위 1% 경계값으로 대체하려 합니다.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, clip, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

굵은 텍스트- 앞 문항 결과 사용 : 2번 `df`
- 아래 안내대로 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df
    - oilPressure 의 1% 분위수를 low, 99% 분위수를 high 로 산출하세요 (변수명: low, high)
    - low 보다 작은 값은 low 로, high 보다 큰 값은 high 로 바꾸세요(행 삭제 없음)
    - 식별자 컬럼 serialNo 를 삭제한 사본에 적용해 df_temp 에 보관하세요
- **[정답05]** 처리 후 oilPressure 의 **최솟값**을 소수점 아래 둘째 자리까지 기입하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요
- 배점 : 6점

---

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)

low, high = df['oilPressure'].quantile([0.01, 0.99])

df_temp = df.drop(columns=['serialNo']).copy()
df_temp['oilPressure'] = df_temp['oilPressure'].clip(lower=low, upper=high)

df_temp['oilPressure'].min()

round(df_temp['oilPressure'].min(), 2)
````
