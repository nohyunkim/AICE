### **5. 평균 체류 시간(분)(SESSION_MIN) 컬럼의 이상치를 IQR 기준으로 제거할 계획입니다. 다음 지침에 따라 처리하십시오.**

- 앞 문항 결과 사용 : 2번 `df_raw`
- 다음 지침에 따라 코드를 작성하고 실행하세요

    - SESSION_MIN 이(가) 두 fence 밖에 있는 행을 삭제하세요
    - 식별자 컬럼 SERIAL_NO 도 삭제하고 인덱스를 0부터 다시 매겨 df_clean 에 보관하세요

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)

q1 = df_raw['SESSION_MIN'].quantile(0.25)
q3 = df_raw['SESSION_MIN'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

is_out = (df_raw['SESSION_MIN'] > upper_fence) | (df_raw['SESSION_MIN'] < lower_fence)
df_clean = df_raw[~is_out].drop(columns=['SERIAL_NO']).reset_index(drop=True)

print(df_clean.shape)
````
