    - IQR 기준(K=2.0)으로 lower_fence, upper_fence 를 계산하세요
    - FE_N_Certs 값이 두 fence 를 벗어나면 가장 가까운 fence 값으로 바꾸세요(행은 삭제하지 않음)

- **[결과 변수]** 결과는 반드시 `data_temp` 변수에 저장하세요 — 변수의 내용(모양·값)으로 채점합니다

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)
q1 = data['FE_N_Certs'].quantile(0.25)
q3 = data['FE_N_Certs'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

data_temp = data.drop(columns=['E_Row_Id']).copy()
data_temp['FE_N_Certs'] = data_temp['FE_N_Certs'].clip(lower=lower_fence, upper=upper_fence)

print(data_temp.shape)
````
