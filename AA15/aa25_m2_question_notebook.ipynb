<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, median, drop) · NumPy(where) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 1번 `np` · 2번 `df`
- 아래 가이드에 따라 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df
    - IQR 기준(K=1.5)으로 lower_fence, upper_fence 를 구하세요
    - 식별자 컬럼 Case_Id 를 삭제한 사본을 df_temp 에 보관하세요
    - df_temp 의 Albumin 중앙값을 `med` 에 저장하고, 두 fence 를 벗어난 값은 `np.where` 로 `med` 로 바꾸세요(행은 삭제하지 않음)
- **[결과 변수]** 결과는 반드시 `df_temp` 변수에 저장하세요 — 변수의 내용(모양·값)으로 채점합니다

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)

q1 = df['Albumin'].quantile(0.25)
q3 = df['Albumin'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

df_temp = df.drop(columns=['Case_Id']).copy()
med = df_temp['Albumin'].median()
df_temp['Albumin'] = np.where((df_temp['Albumin'] < lower_fence) | (df_temp['Albumin'] > upper_fence), med, df_temp['Albumin'])

print(df_temp.shape)
````
