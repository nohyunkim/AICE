### **5. 대출 신청액(만원)(loan_amount) 컬럼의 이상치를 IQR 기준으로 제거할 계획입니다. 제시된 가이드에 맞춰 처리하십시오.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, reset_index, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 제시된 가이드에 맞춰 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df_raw
    - 사분위수(Q1, Q3)와 IQR 을 구하세요
    - lower_fence = Q1 - 3.0 x IQR, upper_fence = Q3 + 3.0 x IQR 로 정의하세요
    - loan_amount 이(가) 두 fence 밖에 있는 행을 삭제하세요
    - 식별자 컬럼 case_id 도 삭제하고 인덱스를 0부터 다시 매겨 df_clean 에 담으세요

````python
# 작업 셀 : 가이드에 따라 코드를 작성하고 실행하세요 (채점하지 않음)

q1 = df_raw['loan_amount'].quantile(0.25)
q3 = df_raw['loan_amount'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 3.0 * iqr
upper_fence = q3 + 3.0 * iqr

is_out = (df_raw['loan_amount'] > upper_fence) | (df_raw['loan_amount'] < lower_fence)
df_clean = df_raw[~is_out].drop(columns=['case_id']).reset_index(drop=True)

print(df_clean.shape)
````
