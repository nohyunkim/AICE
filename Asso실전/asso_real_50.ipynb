<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, median, drop) · NumPy(where) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `data`
- 다음 지침에 따라 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : data
    - IQR 기준(K=1.5)으로 lower_fence, upper_fence 를 산출하세요
    - ② 식별번호 열 삭제 — 식별자 컬럼 rowId 를 삭제한 사본을 data_temp 에 보관하세요
    - data_temp 의 hour 중앙값을 `med` 에 저장하고, 두 fence 를 벗어난 값은 `np.where` 로 `med` 로 바꾸세요(행은 삭제하지 않음)
- **[결과 변수]** 결과는 반드시 `data_temp` 변수에 저장하세요 — 변수의 내용(모양·값)으로 채점합니다
- 확인 : 코드 셀 맨 아래 확인 줄(`print(data_temp.shape)`)을 실행하세요.
- 배점 : 6점

---

````python
# [5번 문항] 여기에 답안코드를 작성하고 실행하세요.

q1 = data['hour'].quantile(0.25)
q3 = data['hour'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

data_temp = data.copy()
data_temp = data_temp.drop('rowId', axis=1)
med = data_temp['hour'].median()

data_temp['hour'] = np.where(
    (data_temp['hour'] < lower_fence) |
    (data_temp['hour'] > upper_fence),
    med,
    data_temp['hour']
)

print(data_temp.shape)
````

fence 를 벗어나는 값을 조건으로 만들고, 그 값만 중앙값으로 바꿉니다. **개념** `quantile(0.25)` 는 1사분위수(Q1), `quantile(0.75)` 는 3사분위수(Q3), IQR = Q3 − Q1 입니다.

    - 출력층: 노드 1개, 활성화 함수 'sigmoid'
    - 모델을 컴파일할 때 optimizer 는 adam, 손실 함수는 binary_crossentropy, 평가지표는 [accuracy] 로 하세요
    - batch_size 는 32, epochs 는 100 로 학습하고 학습정보는 history 변수에 담으세요 (검증 데이터: X_valid, 검증 레이블)
    - 안내하지 않은 파라미터는 지정하지 마세요
- **[결과 변수]** 모델은 `model`, 학습 기록은 `history` 변수에 저장하세요 — 모델 구조(층 수·파라미터 수)와 학습 설정으로 채점합니다
- 확인 : 코드 셀 맨 아래 `print(model.count_params())`을 실행해 확인하세요

````python

# [13번 문항] 여기에 답안코드를 작성하고 실행하세요.

inputs = Input(shape=(X_train.shape[1],))

x = Dense(32, activation='relu')(inputs)
x = Dropout(0.2)(x)
x = Dense(16, activation='relu')(x)

outputs = Dense(1, activation='sigmoid')(x)

model = Model(inputs=inputs, outputs=outputs)

estop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

cb_list = [estop]

cb_list.append(
    ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3
    )
)

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_valid, y_valid),
    batch_size=32,
    epochs=100,
    callbacks=cb_list
)

print(model.count_params())
````
