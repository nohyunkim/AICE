### **5. 최근 승률(%)(winRate) 컬럼의 극단값을 상·하위 1% 경계값으로 대체할 계획입니다.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, clip, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 다음 지침에 따라 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df_raw
    - winRate 의 1% 분위수를 low, 99% 분위수를 high 로 계산하세요 (변수명: low, high)
    - low 보다 작은 값은 low 로, high 보다 큰 값은 high 로 바꾸세요(행 삭제 없음)
    - ② 식별번호 열 삭제 — 식별자 컬럼 rowId 를 삭제한 사본에 적용해 df_clean 에 담으세요
- **[결과 변수]** 결과는 반드시 `df_clean` 변수에 저장하세요 — 변수의 내용(모양·값)으로 채점합니다
- 확인 : 코드 셀 맨 아래 확인 줄(`print(df_clean.shape)`)을 실행하세요.
- 배점 : 6점

---

````python
# [13번 문항] 여기에 답안코드를 작성하고 실행하세요.

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_valid_enc = le.transform(y_valid)

model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(16, activation='relu'),
    Dropout(0.2),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

estop = EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True
)
cb_list = [estop]

model.compile(optimizer='Nadam', loss='binary_crossentropy', metrics=['accuracy'])

history = model.fit(
    X_train,
    y_train_enc,
    epochs=30,
    batch_size=128,
    validation_data=(X_valid, y_valid_enc),
    callbacks=cb_list
)

print(model.count_params())
````
