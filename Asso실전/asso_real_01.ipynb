### **5. 찜 상품 수(wishlist_cnt) 컬럼의 이상치는 삭제하지 않고 정상 범위의 경계값으로 대체할 계획입니다.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, copy, clip, drop) | 난이도: 보통 | 예상 시간: 10분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 다음 지침에 따라 코드를 작성하고 실행하세요
    - 대상 데이터프레임 : df_raw
    - IQR 기준(K=3.0)으로 lower_fence, upper_fence 를 구하세요
    - wishlist_cnt 값이 두 fence 를 벗어나면 가장 가까운 fence 값으로 바꾸세요(행은 삭제하지 않음)
    - ② 식별번호 열 삭제 — 식별자 컬럼 case_id 를 삭제한 사본에 적용해 df_clean 에 보관하세요
- **[결과 변수]** 결과는 반드시 `df_clean` 변수에 저장하세요 — 변수의 내용(모양·값)으로 채점합니다
- 확인 : 코드 셀 맨 아래 확인 줄(`print(df_clean.shape)`)을 실행하세요.
- 배점 : 6점

---

````python
# [5번 문항] 여기에 답안코드를 작성하고 실행하세요.

q1 = df_raw['wishlist_cnt'].quantile(0.25)
q3 = df_raw['wishlist_cnt'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 3.0 * iqr
upper_fence = q3 + 3.0 * iqr

df_clean = df_raw.copy()
df_clean = df_clean.drop('case_id', axis=1)
df_clean['wishlist_cnt'] = df_clean['wishlist_cnt'].clip(lower_fence, upper_fence)

print(df_clean.shape)
````

<details><summary>힌트 1단계 (접근 방향) - 5번 문항</summary>

````python
# [13번 문항] 여기에 답안코드를 작성하고 실행하세요.

inputs = Input(shape=(X_train.shape[1],))
x = Dense(64, activation='elu')(inputs)
outputs = Dense(1, activation='sigmoid')(x)
model = Model(inputs=inputs, outputs=outputs)

estop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
cb_list = [estop]

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=200, batch_size=32, validation_split=0.2, callbacks=cb_list)

print(model.count_params())
````
