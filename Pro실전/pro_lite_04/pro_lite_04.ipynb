````python
# 1. 라이브러리
import numpy as np
import pandas as pd
import joblib
from pathlib import Path

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import f1_score, precision_score, recall_score
from sklearn.base import clone

# 2. 데이터 불러오기
DATA_DIR = Path('/content')

train_path = DATA_DIR / '01_train.csv'
test_path = DATA_DIR / '01_test_x.csv'

if not test_path.exists():
    test_path = DATA_DIR / '01_test_x(1).csv'

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)

print("학습 데이터:", train.shape)
print("테스트 데이터:", test.shape)

print("\n[학습 데이터 정보]")
train.info(verbose=False)

print("\n[기초 통계]")
display(train.describe())

print("\n[결측치]")
print("학습:", train.isnull().sum().sum())
print("테스트:", test.isnull().sum().sum())

print("\n[정답 분포]")
print(train['Attrition'].value_counts())
print(train['Attrition'].value_counts(normalize=True))

# 3. 데이터 전처리 준비
DROP = [
    'id',
    'EmployeeCount',
    'Over18',
    'StandardHours'
]

X = train.drop(columns=DROP + ['Attrition'])
y = train['Attrition'].astype(int)
X_test = test.drop(columns=DROP)

assert X.columns.equals(X_test.columns)

num_cols = X.select_dtypes(include='number').columns.tolist()
cat_cols = X.select_dtypes(include='object').columns.tolist()

print("\n[전처리 대상]")
print("숫자형:", len(num_cols))
print("범주형:", len(cat_cols))

# 4. ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ]
)

# 5. LogisticRegression + Pipeline
base_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LogisticRegression(
            class_weight='balanced',
            max_iter=2000,
            solver='liblinear',
            random_state=42
        ))
    ]
)

# 6. 5-Fold 교차검증
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

candidates = {
    'LR_balanced': clone(base_pipeline),
    'LR_default': clone(base_pipeline).set_params(
        model__class_weight=None
    )
}

cv_probs = {}

print("\n교차검증 결과")

for name, pipeline in candidates.items():

    probs = cross_val_predict(
        pipeline,
        X,
        y,
        cv=skf,
        method='predict_proba'
    )[:, 1]

    cv_probs[name] = probs
    preds = (probs >= 0.5).astype(int)

    print(f"\n[{name}]")
    print("F1       :", round(f1_score(y, preds), 4))
    print("Precision:", round(precision_score(y, preds), 4))
    print("Recall   :", round(recall_score(y, preds), 4))

# 7. 분류 임계값 최적화
thresholds = np.round(np.arange(0.20, 0.71, 0.05), 2)

results = []

for name, probs in cv_probs.items():
    for threshold in thresholds:
        preds = (probs >= threshold).astype(int)

        results.append({
            'Model': name,
            'Threshold': threshold,
            'F1': f1_score(y, preds),
            'Precision': precision_score(y, preds, zero_division=0),
            'Recall': recall_score(y, preds, zero_division=0)
        })

threshold_results = pd.DataFrame(results)

print("\n 임계값별 상위 10개 ")
display(
    threshold_results.sort_values(
        'F1', ascending=False
    ).head(10).round(4)
)

# 8. 기존 만점 모델의 검증된 설정 사용
# LR_default / threshold=0.40
best_threshold_1 = 0.40

candidate_model_1 = clone(candidates['LR_default'])

# 전체 학습 데이터로 최종 학습
candidate_model_1.fit(X, y)

print("\n 최종 모델 ")
print(candidate_model_1.named_steps['model'])
print("임계값:", best_threshold_1)

# 9. 테스트 데이터 예측
candidate_probs_1 = candidate_model_1.predict_proba(
    X_test
)[:, 1]

candidate_pred_1 = (
    candidate_probs_1 >= best_threshold_1
).astype(int)

print("\n[테스트 예측 결과]")
print("전체:", len(candidate_pred_1))
print("이탈(1):", np.sum(candidate_pred_1 == 1))
print("잔류(0):", np.sum(candidate_pred_1 == 0))
print("처음 10개:", candidate_pred_1[:10])

# 10. 제출 CSV 생성
submission = test.copy()
submission['Attrition'] = candidate_pred_1

assert len(submission) == len(test)
assert list(submission.columns) == list(test.columns) + ['Attrition']
assert submission['Attrition'].isin([0, 1]).all()

# 11. 모델 및 CSV 저장
# 기존 만점 파일이 존재하면 review_1 파일로 별도 저장
answer_model = DATA_DIR / 'answer_1.pkl'
answer_csv = DATA_DIR / 'answer_1.csv'

if answer_model.exists() or answer_csv.exists():
    save_prefix = 'review_1'
    print("\n기존 answer_1 파일 보호: 별도 복습 파일 사용")
else:
    save_prefix = 'answer_1'
    print("\n기존 제출 파일 없음: 정식 파일명으로 저장")

model_path = DATA_DIR / f'{save_prefix}.pkl'
csv_path = DATA_DIR / f'{save_prefix}.csv'

model_bundle = {
    'pipeline': candidate_model_1,
    'threshold': best_threshold_1
}

if model_path.exists() or csv_path.exists():
    print("저장 생략: 같은 이름의 파일이 이미 존재합니다.")
else:
    joblib.dump(model_bundle, model_path)
    submission.to_csv(csv_path, index=False)
    print("모델 저장:", model_path.name)
    print("CSV 저장:", csv_path.name)

# 12. 저장 파일 재현성 검증
if model_path.exists() and csv_path.exists():
    loaded_bundle = joblib.load(model_path)
    loaded_csv = pd.read_csv(csv_path)

    loaded_probs = loaded_bundle['pipeline'].predict_proba(
        X_test
    )[:, 1]

    loaded_pred = (
        loaded_probs >= loaded_bundle['threshold']
    ).astype(int)

    assert list(loaded_csv.columns) == list(test.columns) + ['Attrition']
    assert len(loaded_csv) == len(test)
    assert loaded_csv.drop(columns='Attrition').equals(test)

    if np.array_equal(loaded_pred, loaded_csv['Attrition'].to_numpy()):
        print("저장 모델 재현 검증: PASS")
    else:
        print("저장 모델 재현 검증: FAIL (기존 파일 확인 필요)")

# 13. 기존 만점 제출 파일과 비교
if answer_csv.exists():
    original_submission = pd.read_csv(answer_csv)

    same = (
        list(original_submission.columns) == list(submission.columns)
        and original_submission.equals(submission)
    )

    print("\n기존 만점 CSV와 예측 결과 일치:", same)
    if not same:
        print("주의: 기존 만점 제출 파일을 유지하세요.")

print("\n 1번 문제 코드 실행 완료 ")
display(submission.tail())
````

````python

import pandas as pd
from pathlib import Path

# 1. 데이터 불러오기
DATA_DIR = Path('/content')

train_2 = pd.read_csv(DATA_DIR / '02_train.csv')
test_2 = pd.read_csv(DATA_DIR / '02_test_x.csv')

# 2. 데이터 크기 확인
print("[데이터 크기]")
print("학습 데이터:", train_2.shape)
print("테스트 데이터:", test_2.shape)

# 3. 데이터 구조 확인
print("\n[학습 데이터 정보]")
train_2.info()

# 4. 데이터 미리보기
print("\n[학습 데이터 상위 5행]")
display(train_2.head())

print("\n[테스트 데이터 상위 5행]")
display(test_2.head())

# 5. 정답 라벨 분포 확인
print("\n[주제별 데이터 수]")
print(train_2['topic'].value_counts())

# 6. 뉴스 제목 길이 확인
print("\n[뉴스 제목 길이]")
print("평균 길이:", round(train_2['title'].str.len().mean(), 2))
print("최소 길이:", train_2['title'].str.len().min())
print("최대 길이:", train_2['title'].str.len().max())

# 7. 결측치 확인
print("\n[결측치]")
print("학습:", train_2.isnull().sum().sum())
print("테스트:", test_2.isnull().sum().sum())
````

````python

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. 입력 데이터(X)와 정답(y) 분리
X_2 = train_2['title'].astype(str)
y_2 = train_2['topic']

X_test_2 = test_2['title'].astype(str)

# 2. 학습/검증 데이터 분리 (8:2)
X_train_2, X_valid_2, y_train_2, y_valid_2 = train_test_split(
    X_2,
    y_2,
    test_size=0.2,
    random_state=42,
    stratify=y_2
)

# 3. TF-IDF 방식 준비
word_tfidf = TfidfVectorizer(
    analyzer='word',
    ngram_range=(1, 2),
    min_df=2
)

char_tfidf = TfidfVectorizer(
    analyzer='char_wb',
    ngram_range=(2, 5),
    min_df=2,
    sublinear_tf=True
)

# 4. 분리 결과 확인
print("[학습/검증 데이터]")
print("학습:", X_train_2.shape)
print("검증:", X_valid_2.shape)
print("테스트:", X_test_2.shape)

print("\n[검증 데이터 주제별 분포]")
print(y_valid_2.value_counts())

# 5. 문자 n-gram 동작 확인
sample_title = X_train_2.iloc[0]

print("\n[예시 뉴스 제목]")
print(sample_title)

print("\n[문자 n-gram 예시 (처음 15개)]")
print(char_tfidf.build_analyzer()(sample_title)[:15])

print("\n[TF-IDF 설정]")
print("단어:", word_tfidf)
print("문자:", char_tfidf)
````

````python

from sklearn.pipeline import make_pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.base import clone

import pandas as pd

# 1. 모델 후보 4개 구성
models_2 = {
    'Word + NB': make_pipeline(
        clone(word_tfidf),
        MultinomialNB()
    ),
    'Word + SVC': make_pipeline(
        clone(word_tfidf),
        LinearSVC(random_state=42)
    ),
    'Char + NB': make_pipeline(
        clone(char_tfidf),
        MultinomialNB()
    ),
    'Char + SVC': make_pipeline(
        clone(char_tfidf),
        LinearSVC(random_state=42)
    )
}

# 2. 3-Fold 교차검증 설정
cv_2 = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

# 3. 모델별 교차검증
cv_results_2 = []

for name, model in models_2.items():

    print(f"\n[{name}] 교차검증 시작")

    scores = cross_val_score(
        model,
        X_train_2,
        y_train_2,
        cv=cv_2,
        scoring='accuracy',
        n_jobs=1
    )

    print("Fold별 정확도:", scores.round(4))
    print("평균 정확도:", round(scores.mean(), 4))

    cv_results_2.append({
        'Model': name,
        'CV Accuracy': scores.mean(),
        'CV Std': scores.std()
    })

# 4. 교차검증 결과 비교
cv_results_df_2 = pd.DataFrame(cv_results_2)

cv_results_df_2 = cv_results_df_2.sort_values(
    'CV Accuracy',
    ascending=False
)

print("\n 모델별 정확도 비교 ")

display(cv_results_df_2.round(4))

print("\n현재 최고 모델:")
print(cv_results_df_2.iloc[0]['Model'])
````

````python

from sklearn.base import clone
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. 최고 성능 모델 선택
best_model_2 = clone(models_2['Char + SVC'])

# 2. 학습 데이터로 모델 학습
best_model_2.fit(X_train_2, y_train_2)

print("[모델 학습 완료]")
print(best_model_2)

# 3. 검증 데이터 예측
valid_pred_2 = best_model_2.predict(X_valid_2)

# 4. 정확도 평가
valid_acc_2 = accuracy_score(y_valid_2, valid_pred_2)

print("\n[검증 정확도]")
print(f"Accuracy: {valid_acc_2:.4f}")
print(f"Accuracy(%): {valid_acc_2 * 100:.2f}%")

# 5. 클래스별 성능 확인
print("\n[클래스별 분류 성능]")
print(classification_report(
    y_valid_2,
    valid_pred_2,
    digits=4
))

# 6. 혼동행렬 확인
labels_2 = ['경제', '사회', '세계', '정치']

cm_2 = confusion_matrix(
    y_valid_2,
    valid_pred_2,
    labels=labels_2
)

print("\n[혼동행렬: 실제 주제(행) / 예측 주제(열)]")
display(pd.DataFrame(
    cm_2,
    index=labels_2,
    columns=labels_2
))
````

````python

import numpy as np
from sklearn.base import clone

# 1. 검증에서 선택한 모델 복사
final_model_2 = clone(models_2['Char + SVC'])

# 2. 전체 학습 데이터(8,320개)로 학습
final_model_2.fit(X_2, y_2)

print("[전체 데이터 학습 완료]")
print("학습 건수:", len(X_2))
print("최종 모델:", final_model_2)

# 3. 테스트 데이터 예측
test_pred_2 = final_model_2.predict(X_test_2)

# 4. 예측 결과 확인
print("\n[테스트 예측 결과]")
print("전체 예측 건수:", len(test_pred_2))

print("\n[주제별 예측 건수]")
print(pd.Series(test_pred_2).value_counts())

print("\n[처음 10개 예측]")
print(test_pred_2[:10])

# 5. 예측값 유효성 확인
valid_labels_2 = set(y_2.unique())

assert len(test_pred_2) == len(test_2)
assert set(np.unique(test_pred_2)).issubset(valid_labels_2)

print("\n예측 건수 및 라벨 검증: PASS")
````

````python

import zipfile
import pandas as pd
from pathlib import Path
from PIL import Image

# 1. 파일 경로 설정
DATA_DIR_3 = Path('/content')
ZIP_PATH_3 = DATA_DIR_3 / '03_images.zip'

TRAIN_DIR_3 = DATA_DIR_3 / '03_train'
TEST_DIR_3 = DATA_DIR_3 / '03_test'

# 2. 이미지 압축 해제
if not TRAIN_DIR_3.exists() and not TEST_DIR_3.exists():
    with zipfile.ZipFile(ZIP_PATH_3, 'r') as z:
        z.extractall(DATA_DIR_3)
    print("이미지 압축 해제 완료")
else:
    print("기존 이미지 폴더 확인")

# 3. CSV 불러오기
train_3 = pd.read_csv(DATA_DIR_3 / '03_train.csv')
test_3 = pd.read_csv(DATA_DIR_3 / '03_test_x.csv')

# 4. 데이터 크기
print("\n[데이터 크기]")
print("학습 CSV:", train_3.shape)
print("테스트 CSV:", test_3.shape)

# 5. 라벨 분포
print("\n[클래스별 이미지 수]")
print(train_3['label'].value_counts())

# 6. 이미지 파일 수
print("\n[이미지 파일 수]")
print("학습:", len(list(TRAIN_DIR_3.rglob('*.png'))))
print("테스트:", len(list(TEST_DIR_3.glob('*.png'))))

# 7. 이미지 크기 및 색상 확인
sample_path_3 = (
    TRAIN_DIR_3
    / train_3.iloc[0]['label']
    / train_3.iloc[0]['file_name']
)

with Image.open(sample_path_3) as img:
    print("\n[샘플 이미지 정보]")
    print("이미지 크기:", img.size)
    print("색상 모드:", img.mode)

# 8. CSV와 실제 이미지 파일 일치 확인
missing_train_3 = sum(
    not (TRAIN_DIR_3 / row.label / row.file_name).exists()
    for row in train_3.itertuples(index=False)
)

missing_test_3 = sum(
    not (TEST_DIR_3 / name).exists()
    for name in test_3['file_name']
)

print("\n[파일 검증]")
print("학습 이미지 누락:", missing_train_3)
print("테스트 이미지 누락:", missing_test_3)

assert missing_train_3 == 0
assert missing_test_3 == 0

print("이미지 데이터 검증: PASS")
````

````python

import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.utils import image_dataset_from_directory

# 1. 이미지 설정
IMG_SIZE_3 = (64, 64)
BATCH_SIZE_3 = 32
SEED_3 = 42

tf.keras.utils.set_random_seed(SEED_3)

# 2. 공통 데이터 로딩 설정
dataset_kwargs_3 = dict(
    image_size=IMG_SIZE_3,
    batch_size=BATCH_SIZE_3,
    validation_split=0.2,
    seed=SEED_3,
    label_mode='int'
)

# 3. 학습 이미지 불러오기 (80%)
train_ds_3 = image_dataset_from_directory(
    str(TRAIN_DIR_3),
    subset='training',
    **dataset_kwargs_3
)

# 4. 검증 이미지 불러오기 (20%)
valid_ds_3 = image_dataset_from_directory(
    str(TRAIN_DIR_3),
    subset='validation',
    **dataset_kwargs_3
)

# 5. 클래스 이름 확인
CLASSES_3 = train_ds_3.class_names

print("\n[클래스 목록]")
print(CLASSES_3)

assert CLASSES_3 == valid_ds_3.class_names
assert set(CLASSES_3) == set(train_3['label'].unique())

# 6. 데이터를 NumPy 배열로 변환하고 정규화
def to_numpy_3(ds):
    xs, ys = [], []

    for images, labels in ds:
        xs.append(np.asarray(images, dtype=np.float32) / 255.0)
        ys.append(np.asarray(labels, dtype=np.int32))

    return np.concatenate(xs), np.concatenate(ys)

X_train_img_3, y_train_img_3 = to_numpy_3(train_ds_3)
X_valid_img_3, y_valid_img_3 = to_numpy_3(valid_ds_3)

# 7. 데이터 크기 확인
print("\n[데이터 크기]")
print("학습 이미지:", X_train_img_3.shape)
print("학습 라벨:", y_train_img_3.shape)
print("검증 이미지:", X_valid_img_3.shape)
print("검증 라벨:", y_valid_img_3.shape)

# 8. 정규화 확인
print("\n[픽셀값 범위]")
print("최솟값:", X_train_img_3.min())
print("최댓값:", X_train_img_3.max())

# 9. 클래스별 학습/검증 이미지 수
print("\n[학습 데이터 클래스 분포]")
print(pd.Series(
    y_train_img_3
).map(dict(enumerate(CLASSES_3))).value_counts())

print("\n[검증 데이터 클래스 분포]")
print(pd.Series(
    y_valid_img_3
).map(dict(enumerate(CLASSES_3))).value_counts())

# 10. 기본 검증
assert X_train_img_3.shape == (1600, 64, 64, 3)
assert X_valid_img_3.shape == (400, 64, 64, 3)
assert X_train_img_3.min() >= 0
assert X_train_img_3.max() <= 1

print("\n이미지 데이터 준비: PASS")
````

````python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    BatchNormalization,
    GlobalAveragePooling2D,
    Dense,
    Dropout
)
from tensorflow.keras.optimizers import Adam

# 1. CNN 모델 구성
cnn_model_3 = Sequential([
    Input(shape=(64, 64, 3)),

    # 첫 번째 합성곱 블록
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    BatchNormalization(),

    # 두 번째 합성곱 블록
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    BatchNormalization(),

    # 세 번째 합성곱 블록
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    MaxPooling2D((2, 2)),
    BatchNormalization(),

    # 분류기
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.3),

    # 출력층: 5개 클래스
    Dense(len(CLASSES_3), activation='softmax')
])

# 2. 모델 컴파일
cnn_model_3.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 3. 모델 구조 확인
print("[CNN 모델 구조]")
cnn_model_3.summary()

# 4. 출력 클래스 수 검증
assert cnn_model_3.output_shape[-1] == len(CLASSES_3)

print("\nCNN 모델 구성 및 컴파일: PASS")
````

````python

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

# 1. EarlyStopping 설정
early_stop_3 = EarlyStopping(
    monitor='val_loss',
    patience=8,
    min_delta=0.001,
    restore_best_weights=True,
    verbose=1
)

# 2. 학습률 자동 조정
reduce_lr_3 = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=4,
    min_lr=0.00001,
    verbose=1
)

# 3. CNN 실제 학습
history_3 = cnn_model_3.fit(
    X_train_img_3,
    y_train_img_3,
    validation_data=(
        X_valid_img_3,
        y_valid_img_3
    ),
    epochs=40,
    batch_size=32,
    callbacks=[
        early_stop_3,
        reduce_lr_3
    ],
    verbose=1
)

# 4. 학습 결과 확인
print("\n CNN 학습 완료 ")

print("실제 학습 Epoch 수:", len(history_3.history['loss']))

# 5. 최저 검증 손실 확인
best_epoch_3 = (
    int(np.argmin(history_3.history['val_loss'])) + 1
)

print("검증 손실 최저 Epoch:", best_epoch_3)
print(
    "최저 검증 Loss:",
    round(min(history_3.history['val_loss']), 4)
)

# 6. 조기 종료 후 복원된 모델 성능 평가
val_loss_3, val_acc_3 = cnn_model_3.evaluate(
    X_valid_img_3,
    y_valid_img_3,
    verbose=0
)

print("\n[최종 검증 성능]")
print(f"Validation Loss    : {val_loss_3:.4f}")
print(f"Validation Accuracy: {val_acc_3:.4f}")
print(f"Validation Accuracy(%): {val_acc_3 * 100:.2f}%")
````

````python
# 전체 자동 채점 — 세 문항의 제출 파일(answer_1~3.csv·모델 파일)을 만든 뒤 이 셀을 실행하세요.
# 정답(테스트 라벨)은 압축·인코딩되어 있어 보이지 않습니다. 문항별 지표·통과 여부와 총점을 출력합니다.
PREFIX = 'answer'                                  # 제출 파일명 앞부분(고정)

import base64, json, os, re, zlib
import numpy as np
import pandas as pd

_KEY = "IoCXBpc0gRxKJE8ZvXpK904UIbEpXtwakS4aHvxW03QwXuROY6wA+nIgklmujgSAGQVyKoX7/kYC6ObGp+Sj3BR07B53zelhpYWWqKe+MOdgAQzNwZMtHTWEFcAPreUiK27jAI8XgQXGF+E1kaj2ACwFfIXtt+A0sQ9ps/QEDIfNHyWj+LLl7Y/8xwWOkfDlbne8gbVl5AY2JLIlJw2Hlk3xt3avoywi4yQB74ets5EN9aEdzMGXJgNtj6f/gystt/x9MejHJ609oKJVIJekGZcIiglXVLtzL6Rz5JDz5E7LX6nu2OP5qpNbqjFL5tMFU0Ejx3SogQ7ytkQ1EdB78NBrLOYijP9kw7WurJ+LPe0oppL3Jum1DICPD70SROPOKOviBM3Rwf3Uv8Iqdn0VR5GTLr0CtGSkKK2V9adnLUWkpK6N3bf9h++H/cWkorXtxV2d/yQXJyBCrbOVJbVkpCRqETgGqDN8XyFptTyDJyDiOTVGEgW8ALAz072AYe7BdR8FqDNzNP8saOvsjiaTPf5yEmfZzM2WJuxu+6mryDe0wGd04w1y1QoQSeWYyMIi3PfXQxMeIz71V0bN+9+M2lZNFqcXlDipWc7Z1gSAqD0x+edB7I7bTH7pBAExF2OSCjAx0PdR5s7aIMO7TGBUzleCvIKOwcRTzGXJ2UNRmBHHsxk3B98tkvxOh3GOAMLMW+yDctVOk/z/R+zH+nccuRQkRvYLrIRxrqvkAar8rJBH63kyxcmx6p2DpJmGiZZH8l/8d0Y25yPTGq+ZJHTyUcg2+qtFMBsa5KtAupICIgFWAEBufC6iZsk8f9Ebad0WPNcAZw65XxCLxlPX/juGmI4AhHhj+F1uxRxpem4l+edbwF36gFm1YrUrCkSrINOcEtjOqkNFdbnyzpojMhJUXL3GcoUytgrTyqPSjNvx+usDEp5XPWs6Dk/XKJiEQ849ZC28C2AOhaIGRpwEo8Lc9GP6iPMeDzKWtRqboqZG75A6nFgxd7BZGsfcZ2s157Nmmut4QCMo9g3zVvqiRdDi+BJdAJUZuSwA9DhKIG7xLh71Otx7e79Q3FggyNasUuxT9I1wAnTV8PvrxLjHi/bI9ZCqYUTYN5isRNbjzvBWAH8ZL0AfpjvDHst4ihxioHKi5+2jm7mm5oychtEg5ULV1wBi5I3fncAZ4BCiBobbw4kPd4I1q4gUHVrUIhmaLaPGhwsz+jlbQiFaqVVz3UruS7fUGyke4nXXHCoLUZ15i0O0+TJOBQtO8i0uKF6NTNnoXoBg+PLpzvHbjJ2LxRKtA5jDMWrQt9K8srbU1cWjBBSC+FHL9fmJSfeTi+C68qr4sv+KY58dJzLskh0FUBRZeZ0mzB+t4/vu+gi4svm/ObpCkXoib0Ilu6G6GXAJyBdeuVddcHMSazalOCBdM15I9+qrx7PUNCVKhc/qMeTZ3rWmvCdEsN8KtsD0+hC9UjK2iosP2Axoo9B9AlkosktwFy6fCbNb3XkFG8mjUg6wzb2/Q69c7sxzSFEJUIyDu1J1/q2XS6vaLMb1gfnonA2UOtbdnQCyUPVOgYgHIIiDCKQec4UpGB0zECcKkKfeSWiI4srpyxDlOQq8T+i5VFAi7ssdDTMhX8VAvrdPBcX740+JMXDHe1XzDqnPT85wpG7dCNPJimZAY8sU2qgPaEBJviP+rlr7YnwluHFyYbtpL45kifJAlXcdL0m/f3Wf5kTbxht3OwqFYAe/VPCOcjVbChS3qb3EikSGwgBuRZtxr/fC6f8mHUQlDNz1Z4tkBBh8GgG4GhNeFhvrwqpjers259iicf16fl7ZHDb+MGclkNVldR4tE3Pn/FRG1PRHZZ0BBUb9I8+m1yVZWQKvv2G92RQscDUE+JX+mUh7jgnOApo76aNg7Ijj/caYvBUH1K1sSEQbUWP1sZoPIl7SrZ57NQTIlEuGpeCtMJ2DqvFHtWXa9SzGV+WA60VpNZCRsLpX1YWUxOa7q6+2vrtX1U3xL205JRWHxgU1YEiQv6a24ATEjZeB1y8nLb38cM2/4VBBhqf1lfH8b9y+M4AhuDQqg26Nh5qfIC8vajyqw9ll6sGbBF3stjpjbvMrbbZLIRpOf0GSptouqpa+bQHIR9oOwF+SgMgQy43/2f99A212W6OjCe6qRitqchVK539+QdztN6pT8IxVZSgqcAr9W8RjRHRyERSus1MLRKuRrTPTF1WQFd5wHvRjwUS8A53FcgBvI7nMhoCbXaG3AILMK8hZslGQRu1zXDSORjvC7Qu3buhWZevQua89QuoTKZ7qxgcZMrgeRft2lPPTmfJmFKuy5oPXRE3O9D42oYLDGi40lIm/ruEYHLdS4ZnZlX6Bo/oC61JD4E48gYGQfU5UEbU66NlJEKy1UjXm5iYMVhKDycLOfHQ5L55RKyHQEnzijuoC1lQNr/bO6FUUHwo2SkfE0GvMSCBZxHlCTNNQrP8qIG5PjwsSZmD6Pm6SpZ7fxRRhGvfSUZyiBmc9KxxIJ19qKIIBg4PSrBfTReLqzQmQy+JhjDTfMtQdEBz2H3MyeNVCkhxUkBTO7wjYVaFwgJJihgR8OLmHWQUuU7QfoqazrqyurCVWX7rb"


def _key():
    raw = base64.b64decode(_KEY)
    return json.loads(zlib.decompress(bytes(b ^ 0x5A for b in raw)).decode("utf-8"))


def _metric(m, y, p):
    if m == "rmse":
        return float(np.sqrt(np.mean((np.asarray(p, float) - np.asarray(y, float)) ** 2)))
    y, p = np.asarray(y).astype(str), np.asarray(p).astype(str)
    if m == "accuracy":
        return float((y == p).mean())
    labels = sorted(set(y)) if m == "macro_f1" else ["1"]
    f1s = []
    for c in labels:
        tp = np.sum((p == c) & (y == c)); fp = np.sum((p == c) & (y != c)); fn = np.sum((p != c) & (y == c))
        f1s.append(0.0 if tp == 0 else 2 * tp / (2 * tp + fp + fn))
    return float(np.mean(f1s))


def grade(prefix="answer", folder="."):
    key = _key()
    phone = prefix                                      # 제출 파일명 앞부분(고정: answer_N.csv · answer_N.pkl/.h5)
    total, rows = 0, []
    for q in key["problems"]:
        no, pts = q["no"], q["points"]
        csv = os.path.join(folder, f"{phone}_{no}.csv")
        models = [f for f in os.listdir(folder) if f.startswith(f"{phone}_{no}.") and not f.endswith(".csv")]
        msg, score, val = [], 0, None
        if not models:
            msg.append("모델 파일 없음")
        if not os.path.exists(csv):
            rows.append((no, pts, 0, None, "예측 CSV 없음")); continue
        sub = pd.read_csv(csv)
        test = pd.read_csv(os.path.join(folder, q["test_file"]))
        lab = q["label"]
        if list(sub.columns) != list(test.columns) + [lab]:
            msg.append(f"열 구조가 다름(테스트 열 + 마지막 '{lab}' 열이어야 함)")
        if lab not in sub.columns or len(sub) != len(test):
            rows.append((no, pts, 0, None, f"행 수 {len(sub)} ≠ {len(test)} 또는 '{lab}' 열 없음")); continue
        keycol = q["key_col"]
        if keycol and not (sub[keycol].astype(str).values == test[keycol].astype(str).values).all():
            msg.append(f"'{keycol}' 순서가 테스트 파일과 다름")
        if sub[lab].isna().any():
            msg.append("예측값에 결측 있음"); sub[lab] = sub[lab].fillna(0 if q["metric"] == "rmse" else "")
        val = _metric(q["metric"], q["y"], sub[lab].values)
        ok = val <= q["pass_line"] if q["metric"] == "rmse" else val >= q["pass_line"]
        if ok and not msg:
            score = pts
        rows.append((no, pts, score, val, "; ".join(msg) or ("통과" if ok else "기준선 미달")))
        total += score
    print(f"{'문항':<6}{'배점':>4}{'점수':>6}  {'지표':<10}{'결과':>10}  {'목표':>8}  {'기준선':>8}  비고")
    for (no, pts, sc, val, note), q in zip(rows, key["problems"]):
        v = "-" if val is None else (f"{val:.4f}")
        print(f"{no:<6}{pts:>4}{sc:>6}  {q['metric']:<10}{v:>10}  {q['target']:>8}  {q['pass_line']:>8}  {note}")
    print(f"총점: {total} / 100  ({'합격' if total >= 80 else '불합격'}, 합격 기준 80점)")
    return total


_점수 = grade(prefix=PREFIX)
````
