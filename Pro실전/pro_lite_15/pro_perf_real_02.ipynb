````python

import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

# 1. 실제 업로드 파일 찾기
def find_csv(prefix):
    for folder in [Path("."), Path("/mnt/data")]:
        exact = folder / f"{prefix}.csv"
        if exact.exists():
            return str(exact)
        matches = sorted(folder.glob(f"{prefix}*.csv"))
        if matches:
            return str(matches[0])
    raise FileNotFoundError(f"{prefix} 파일을 찾을 수 없습니다.")

train = pd.read_csv(find_csv("01_train"))
test = pd.read_csv(find_csv("01_test_x"))
LABEL = "CWT값"

print("훈련:", train.shape, "/ 테스트:", test.shape)
print("기간:", train["평가월"].min(), "~", train["평가월"].max())
print("결측값:", train.isna().sum().sum())
print(train[LABEL].describe().round(2))

# 2. 파생변수 및 범주형 인코딩
def feats(df):
    d = df.drop(columns=["id", "평가월", LABEL],
                errors="ignore").copy()

    cwt = [f"CWT_lag{i}" for i in range(1, 7)]
    rwt = [f"RWT_lag{i}" for i in range(1, 4)]

    d["CWT_3평균"] = df[cwt[:3]].mean(axis=1)
    d["CWT_6평균"] = df[cwt].mean(axis=1)
    d["CWT_3표준편차"] = df[cwt[:3]].std(axis=1)
    d["CWT_단기추세"] = df["CWT_lag1"] - df["CWT_lag3"]
    d["CWT_전월변화"] = df["CWT_lag1"] - df["CWT_lag2"]

    d["RWT_3평균"] = df[rwt].mean(axis=1)
    d["RWT_추세"] = df["RWT_lag1"] - df["RWT_lag3"]
    d["CWT_RWT_차이"] = df["RWT_lag1"] - df["CWT_lag1"]

    d["훈련월"] = df["월"].isin([3, 10]).astype(int)
    d["훈련직후"] = df["월"].isin([4, 11]).astype(int)
    d["분기"] = ((df["월"] - 1) // 3 + 1).astype(str)

    return pd.get_dummies(
        d, columns=["부대코드", "종별", "월", "분기"],
        dtype=int
    )

X = feats(train)
y = train[LABEL]

# 3. 시간순 검증 (마지막 6개월)
mask = train["평가월"] < "2025-01"
X_train, X_valid = X.loc[mask], X.loc[~mask]
y_train, y_valid = y.loc[mask], y.loc[~mask]

print("학습:", len(X_train), "/ 검증:", len(X_valid))

# 4. 모델 학습
model_1 = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
)
model_1.fit(X_train, y_train)

# 5. RMSE 검증
pred = model_1.predict(X_valid)
rmse = np.sqrt(mean_squared_error(y_valid, pred))

print(f"\n검증 RMSE: {rmse:.4f}")
print("달성 목표: RMSE <= 3.096")
print("결과:", "목표 달성" if rmse <= 3.096 else "추가 개선 필요")
````

````python

import pandas as pd
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. 데이터 불러오기
def find_csv(prefix):
    for folder in [Path("."), Path("/mnt/data")]:
        matches = sorted(folder.glob(f"{prefix}*.csv"))
        if matches:
            return str(matches[0])
    raise FileNotFoundError(prefix)

train_2 = pd.read_csv(find_csv("02_train"))
test_2 = pd.read_csv(find_csv("02_test_x"))

LABEL_2 = "심사결과"

print("훈련:", train_2.shape, "/ 테스트:", test_2.shape)
print("정답 분포:\n", train_2[LABEL_2].value_counts())
print("결측값:", train_2.isna().sum().sum())
print("평균 문장 길이:", round(train_2["기고문요약"].str.len().mean(), 1))

# 2. 입력 데이터 구성
X_2 = train_2.drop(columns=["id", LABEL_2])
y_2 = train_2[LABEL_2]

numeric_cols = [
    "유사도점수", "표절의심여부",
    "선행연구건수", "참고문헌수",
    "분량쪽", "심사소요시간분"
]

preprocessor_2 = ColumnTransformer([
    ("text", TfidfVectorizer(
        analyzer="char",
        ngram_range=(2, 5),
        min_df=2,
        max_features=15000,
        sublinear_tf=True
    ), "기고문요약"),

    ("num", StandardScaler(), numeric_cols),

    ("cat", OneHotEncoder(handle_unknown="ignore"), ["분야"])
])

# 3. 모델 구성
model_2 = Pipeline([
    ("preprocess", preprocessor_2),
    ("classifier", LogisticRegression(
        C=2.0,
        max_iter=2000,
        random_state=42
    ))
])

# 4. 학습·검증 분리
X_train_2, X_valid_2, y_train_2, y_valid_2 = train_test_split(
    X_2, y_2,
    test_size=0.2,
    random_state=42,
    stratify=y_2
)

# 5. 학습 및 평가
model_2.fit(X_train_2, y_train_2)
pred_2 = model_2.predict(X_valid_2)

acc_2 = accuracy_score(y_valid_2, pred_2)

print(f"\n검증 Accuracy: {acc_2:.4%}")
print("달성 목표: 70%")
print("결과:", "✅ 목표 달성" if acc_2 >= 0.70 else "❌ 추가 개선 필요")

print("\n클래스별 평가:")
print(classification_report(y_valid_2, pred_2))
````

````python

import os, zipfile
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    BatchNormalization, Flatten, Dense, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. 파일 찾기
def find_file(prefix, ext):
    for folder in [Path("."), Path("/mnt/data")]:
        files = sorted(folder.glob(f"{prefix}*.{ext}"))
        if files:
            return files[0]
    raise FileNotFoundError(prefix)

train_3 = pd.read_csv(find_file("03_train", "csv"))
test_3 = pd.read_csv(find_file("03_test_x", "csv"))

zip_path = find_file("03_images", "zip")
IMG_DIR = Path(os.environ.get("AICE_IMG_DIR", "aice_images"))

if not (IMG_DIR / "03_train").exists():
    IMG_DIR.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(IMG_DIR)

# 2. 이미지 및 라벨 준비
SIZE = 64
CLASSES_3 = sorted(train_3["label"].unique())
label_to_idx = {v: i for i, v in enumerate(CLASSES_3)}

def load_image(path):
    with Image.open(path) as im:
        return np.asarray(
            im.convert("RGB").resize((SIZE, SIZE)),
            dtype=np.float32
        ) / 255.0

X_3 = np.stack([
    load_image(IMG_DIR / "03_train" / label / filename)
    for filename, label in zip(
        train_3["file_name"], train_3["label"]
    )
])

y_3 = train_3["label"].map(label_to_idx).to_numpy()

print("훈련:", X_3.shape, "/ 테스트:", test_3.shape)
print("클래스:", CLASSES_3)
print("결측:", train_3.isna().sum().sum())

# 3. 학습·검증 분리
X_tr3, X_val3, y_tr3, y_val3 = train_test_split(
    X_3, y_3,
    test_size=0.2,
    random_state=42,
    stratify=y_3
)

# 4. 데이터 증강 (좌우 반전 제외)
augment_3 = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.07,
    height_shift_range=0.07,
    zoom_range=0.10,
    fill_mode="nearest"
)

# 5. CNN 모델
np.random.seed(42)
tf.random.set_seed(42)

model_3 = Sequential([
    Input(shape=(64, 64, 3)),

    Conv2D(32, 3, activation="relu", padding="same"),
    Conv2D(32, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Conv2D(64, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Conv2D(128, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.4),
    Dense(len(CLASSES_3), activation="softmax")
])

model_3.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 6. 학습
callbacks_3 = [
    EarlyStopping(
        monitor="val_accuracy",
        patience=7,
        restore_best_weights=True,
        mode="max"
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=0.00001
    )
]

history_3 = model_3.fit(
    augment_3.flow(X_tr3, y_tr3, batch_size=32, seed=42),
    validation_data=(X_val3, y_val3),
    epochs=30,
    callbacks=callbacks_3,
    verbose=1
)

# 7. 검증
pred_3 = np.argmax(model_3.predict(X_val3, verbose=0), axis=1)
acc_3 = accuracy_score(y_val3, pred_3)

print(f"\n검증 Accuracy: {acc_3:.4%}")
print("달성 목표: 88%")
print("결과:", "목표 달성" if acc_3 >= 0.88 else "추가 개선 필요")
print("\n클래스별 평가:")
print(classification_report(
    y_val3, pred_3,
    target_names=CLASSES_3,
    zero_division=0
))
````

````python
import os, shutil
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    BatchNormalization, Flatten, Dense, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import tensorflow as tf

# 1. 테스트 이미지 경로 찾기 함수
def find_test_image_root(base_dir):
    candidates = [
        base_dir / "03_test",
        base_dir / "test",
        base_dir / "03_images" / "03_test",
        base_dir
    ]
    for c in candidates:
        if c.exists():
            sample_ok = sum((c / fn).exists() for fn in test_3["file_name"].head(10))
            if sample_ok >= 3:
                return c
    raise FileNotFoundError("테스트 이미지 폴더를 찾지 못했습니다.")

TEST_IMG_DIR = find_test_image_root(IMG_DIR)
print("테스트 이미지 폴더:", TEST_IMG_DIR)

# 2. 전체 훈련 이미지 다시 준비
X_full_3 = X_3
y_full_3 = y_3

# 3. 데이터 증강
augment_full_3 = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.07,
    height_shift_range=0.07,
    zoom_range=0.10,
    fill_mode="nearest",
    validation_split=0.1
)

# 4. 최종 CNN 재구성
np.random.seed(42)
tf.random.set_seed(42)

final_model_3 = Sequential([
    Input(shape=(64, 64, 3)),

    Conv2D(32, 3, activation="relu", padding="same"),
    Conv2D(32, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Conv2D(64, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Conv2D(128, 3, activation="relu", padding="same"),
    MaxPooling2D(),
    BatchNormalization(),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.4),
    Dense(len(CLASSES_3), activation="softmax")
])

final_model_3.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

best_tmp = "best_answer_3.keras"

callbacks_final_3 = [
    EarlyStopping(
        monitor="val_accuracy",
        patience=6,
        restore_best_weights=True,
        mode="max"
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-5
    ),
    ModelCheckpoint(
        best_tmp,
        monitor="val_accuracy",
        save_best_only=True,
        mode="max"
    )
]

# 5. 전체 데이터 학습
history_final_3 = final_model_3.fit(
    augment_full_3.flow(
        X_full_3, y_full_3,
        batch_size=32,
        subset="training",
        seed=42
    ),
    validation_data=augment_full_3.flow(
        X_full_3, y_full_3,
        batch_size=32,
        subset="validation",
        shuffle=False
    ),
    epochs=25,
    callbacks=callbacks_final_3,
    verbose=1
)

# 6. 최고 성능 모델을 answer_3.h5로 저장
best_model_3 = load_model(best_tmp)
best_model_3.save("answer_3.h5")

# 7. 테스트 이미지 로드 및 예측
def load_test_image(path):
    with Image.open(path) as im:
        return np.asarray(
            im.convert("RGB").resize((SIZE, SIZE)),
            dtype=np.float32
        ) / 255.0

X_test_3 = np.stack([
    load_test_image(TEST_IMG_DIR / fn)
    for fn in test_3["file_name"]
])

pred_prob_before_3 = best_model_3.predict(X_test_3, verbose=0)
pred_idx_before_3 = np.argmax(pred_prob_before_3, axis=1)
pred_label_before_3 = [CLASSES_3[i] for i in pred_idx_before_3]

# 8. 저장 모델 복원 후 예측 일치 확인
loaded_3 = load_model("answer_3.h5")
pred_prob_after_3 = loaded_3.predict(X_test_3, verbose=0)
pred_idx_after_3 = np.argmax(pred_prob_after_3, axis=1)
pred_label_after_3 = [CLASSES_3[i] for i in pred_idx_after_3]

# 9. 제출 CSV 생성
answer_3 = test_3.copy()
answer_3["label"] = pred_label_after_3
answer_3.to_csv("answer_3.csv", index=False)

# 10. 제출 검증
saved_3 = pd.read_csv("answer_3.csv")

assert Path("answer_3.h5").exists()
assert Path("answer_3.csv").exists()
assert len(saved_3) == len(test_3)
assert list(saved_3.columns) == list(test_3.columns) + ["label"]

pd.testing.assert_frame_equal(
    saved_3[test_3.columns],
    test_3,
    check_dtype=False
)

assert saved_3["label"].notna().all()
assert set(saved_3["label"]).issubset(set(CLASSES_3))
assert pred_label_before_3 == pred_label_after_3
assert saved_3["label"].tolist() == pred_label_after_3

print("3번 제출 검증 완료")
print("모델: answer_3.h5")
print("예측: answer_3.csv")
print("제출 형태:", saved_3.shape)
print("\n예측 클래스 분포:")
print(saved_3["label"].value_counts())

# 11. 임시 파일 정리
if Path(best_tmp).exists():
    Path(best_tmp).unlink()

print("\n모델 복원 및 예측값 일치 확인")
````
