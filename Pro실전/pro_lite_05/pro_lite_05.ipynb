````python

from pathlib import Path
import pandas as pd

# 1. 파일 경로 확인
BASE = Path("/content")

train_path = BASE / "01_train.csv"

# 업로드된 파일명에 (1)이 붙은 경우도 처리
test_candidates = [
    BASE / "01_test_x(1).csv",
    BASE / "01_test_x.csv"
]

test_path = next(
    (p for p in test_candidates if p.exists()),
    None
)

if not train_path.exists() or test_path is None:
    raise FileNotFoundError(
        "01_train.csv 또는 01_test_x.csv 파일이 /content에 있는지 확인하세요."
    )

# 2. 데이터 불러오기
train1 = pd.read_csv(train_path)
test1 = pd.read_csv(test_path)

print(" 사용한 파일 ")
print("학습:", train_path.name)
print("테스트:", test_path.name)

# 3. 데이터 크기
print("\n 데이터 크기 ")
print("학습:", train1.shape)
print("테스트:", test1.shape)

# 4. 컬럼 및 자료형
print("\n 학습 데이터 정보 ")
train1.info()

print("\n 테스트 컬럼 ")
print(test1.columns.tolist())

# 5. 실제 데이터 확인
print("\n 학습 데이터 상위 5행 ")
display(train1.head())

print("\n 테스트 데이터 상위 3행")
display(test1.head(3))

# 6. 결측치
print("\n 결측치 ")
print("학습 결측치:", train1.isna().sum().sum())
print("테스트 결측치:", test1.isna().sum().sum())

# 7. 예측 대상(qty) 분포
print("\n qty 통계 ")
print(train1["qty"].describe())

print("\n주문량 0 비율:",
      round((train1["qty"] == 0).mean() * 100, 2), "%")

# 8. 제품 및 날짜 확인
print("\n 제품과 날짜 ")
print("제품 종류:", train1["product"].nunique())
print("학습 기간:", train1["Date"].min(), "~", train1["Date"].max())
print("테스트 기간:", test1["Date"].min(), "~", test1["Date"].max())
````

````python

# 과거 14일 주문량 컬럼 생성
W = 14
LAGS = [f"lag_{k}" for k in range(W, 0, -1)]

# 제품별 날짜순 정렬
train1_lag = train1.sort_values(["product", "Date"]).copy()

# 제품별 과거 주문량 생성
for k in range(1, W + 1):
    train1_lag[f"lag_{k}"] = (
        train1_lag.groupby("product")["qty"].shift(k)
    )

# 과거 14일 기록이 없는 행 제거
train1_lag = train1_lag.dropna(subset=LAGS).reset_index(drop=True)

print("가공 전:", train1.shape)
print("가공 후:", train1_lag.shape)

print("\n생성된 lag 컬럼")
print(LAGS)

print("\n가공 데이터 예시")
display(train1_lag[["Date", "product", "qty", "lag_1", "lag_2", "lag_14"]].head())

print("\nlag 결측치:", train1_lag[LAGS].isna().sum().sum())
print("테스트와 lag 컬럼 일치:", all(c in test1.columns for c in LAGS))
````

````python

# 특징 생성 함수
def add_features(df):
    d = df.copy()

    d["Date"] = pd.to_datetime(d["Date"])
    d["weekday"] = d["Date"].dt.weekday
    d["month"] = d["Date"].dt.month

    recent7 = [f"lag_{k}" for k in range(1, 8)]

    d["mean_7"] = d[recent7].mean(axis=1)
    d["mean_14"] = d[LAGS].mean(axis=1)
    d["max_14"] = d[LAGS].max(axis=1)
    d["zero_14"] = (d[LAGS] == 0).sum(axis=1)

    return d

# 학습과 테스트에 동일하게 적용
train1_feat = add_features(train1_lag)
test1_feat = add_features(test1)

print("학습 데이터:", train1_feat.shape)
print("테스트 데이터:", test1_feat.shape)

print("\n추가된 특징 확인")
cols = ["Date", "product", "qty", "mean_7",
        "mean_14", "max_14", "zero_14", "weekday"]

display(train1_feat[cols].head())

print("\n테스트 행 순서 유지:",
      test1_feat["id"].tolist() == test1["id"].tolist())

print("추가 특징 결측치:",
      train1_feat.isna().sum().sum(),
      test1_feat.isna().sum().sum())
````

````python

# 마지막 30일을 검증 기간으로 설정
cutoff = train1_feat["Date"].max() - pd.Timedelta(days=29)

# 학습·검증 데이터 분리
train1_fit = train1_feat[train1_feat["Date"] < cutoff].copy()
valid1 = train1_feat[train1_feat["Date"] >= cutoff].copy()

print("학습 데이터:", train1_fit.shape)
print("검증 데이터:", valid1.shape)

print("\n학습 기간")
print(train1_fit["Date"].min(), "~", train1_fit["Date"].max())

print("\n검증 기간")
print(valid1["Date"].min(), "~", valid1["Date"].max())

print("\n데이터 분리 확인")
print("행 수 일치:", len(train1_fit) + len(valid1) == len(train1_feat))
print("기간 겹침 없음:", train1_fit["Date"].max() < valid1["Date"].min())
print("검증 제품 종류:", valid1["product"].nunique())
````

````python

import numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

# 사용할 특징 선택
cat_cols = ["product", "group", "sub_group"]
num_cols = LAGS + [
    "mean_7", "mean_14", "max_14",
    "zero_14", "weekday", "month"
]

feature_cols = cat_cols + num_cols

# 범주형 데이터를 숫자로 변환
X_train = pd.get_dummies(
    train1_fit[feature_cols], columns=cat_cols
)
X_valid = pd.get_dummies(
    valid1[feature_cols], columns=cat_cols
)

# 학습과 검증의 컬럼 일치
X_valid = X_valid.reindex(
    columns=X_train.columns, fill_value=0
)

y_train = train1_fit["qty"]
y_valid = valid1["qty"]

# Ridge 모델 학습
ridge1 = Ridge(alpha=50)
ridge1.fit(X_train, y_train)

# 검증 데이터 예측
pred_ridge1 = ridge1.predict(X_valid)
pred_ridge1 = np.clip(pred_ridge1, 0, None)

# 성능 평가
mse_ridge1 = mean_squared_error(y_valid, pred_ridge1)
rmse_ridge1 = np.sqrt(mse_ridge1)

print("학습 특징 개수:", X_train.shape[1])
print("검증 데이터 개수:", len(y_valid))
print("MSE:", round(mse_ridge1, 4))
print("RMSE:", round(rmse_ridge1, 4))
print("목표 RMSE 933.1 이하:", rmse_ridge1 <= 933.1)
````

````python

from lightgbm import LGBMRegressor, early_stopping

# LightGBM 모델 생성
lgbm1 = LGBMRegressor(
    n_estimators=1200,
    learning_rate=0.03,
    num_leaves=15,
    min_child_samples=25,
    subsample=0.85,
    subsample_freq=1,
    colsample_bytree=0.9,
    reg_lambda=3,
    random_state=42,
    verbosity=-1
)

# 모델 학습
lgbm1.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    eval_metric="rmse",
    callbacks=[
        early_stopping(stopping_rounds=80, verbose=False)
    ]
)

# 검증 데이터 예측
pred_lgbm1 = lgbm1.predict(X_valid)
pred_lgbm1 = np.clip(pred_lgbm1, 0, None)

# 성능 평가
mse_lgbm1 = mean_squared_error(y_valid, pred_lgbm1)
rmse_lgbm1 = np.sqrt(mse_lgbm1)

print("최적 학습 횟수:", lgbm1.best_iteration_)
print("MSE:", round(mse_lgbm1, 4))
print("RMSE:", round(rmse_lgbm1, 4))

print("\n모델 성능 비교")
print("Ridge RMSE:", round(rmse_ridge1, 4))
print("LightGBM RMSE:", round(rmse_lgbm1, 4))

print("\n목표 RMSE 933.1 이하:", rmse_lgbm1 <= 933.1)
print("자동 채점 기준 1026.41 이하:", rmse_lgbm1 <= 1026.41)
````

````python

from xgboost import XGBRegressor

# XGBoost 모델 생성
xgb1 = XGBRegressor(
    n_estimators=1500,
    max_depth=5,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.85,
    min_child_weight=5,
    reg_lambda=5,
    objective="reg:squarederror",
    eval_metric="rmse",
    early_stopping_rounds=80,
    random_state=42,
    n_jobs=-1,
    tree_method="hist"
)

# 모델 학습
xgb1.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)

# 검증 데이터 예측
pred_xgb1 = xgb1.predict(X_valid)
pred_xgb1 = np.clip(pred_xgb1, 0, None)

# 성능 평가
mse_xgb1 = mean_squared_error(y_valid, pred_xgb1)
rmse_xgb1 = np.sqrt(mse_xgb1)

print("최적 학습 횟수:", xgb1.best_iteration + 1)
print("MSE:", round(mse_xgb1, 4))
print("RMSE:", round(rmse_xgb1, 4))

print("\n모델 성능 비교")
print("Ridge:", round(rmse_ridge1, 4))
print("LightGBM:", round(rmse_lgbm1, 4))
print("XGBoost:", round(rmse_xgb1, 4))

print("\n목표 RMSE 933.1 이하:", rmse_xgb1 <= 933.1)
print("자동 채점 기준 1026.41 이하:", rmse_xgb1 <= 1026.41)
````

````python

import joblib
from pathlib import Path

model_path = Path("/content/candidate_1.pkl")
csv_path = Path("/content/candidate_1.csv")

# 기존 후보 파일 덮어쓰기 방지
if model_path.exists() or csv_path.exists():
    raise FileExistsError("기존 candidate_1 파일이 있습니다.")

# 전체 학습 데이터 준비
X_full = pd.get_dummies(
    train1_feat[feature_cols], columns=cat_cols
)
y_full = train1_feat["qty"]

# 테스트 데이터 준비
X_test = pd.get_dummies(
    test1_feat[feature_cols], columns=cat_cols
)
X_test = X_test.reindex(
    columns=X_full.columns, fill_value=0
)

# 검증에서 선택한 학습 횟수 적용
params = xgb1.get_params()
params["n_estimators"] = xgb1.best_iteration + 1
params["early_stopping_rounds"] = None

# 전체 데이터로 최종 후보 학습
xgb1_candidate = XGBRegressor(**params)
xgb1_candidate.fit(X_full, y_full)

# 테스트 데이터 예측
pred_candidate1 = xgb1_candidate.predict(X_test)
pred_candidate1 = np.clip(pred_candidate1, 0, None)

# 제출 형식의 후보 CSV 생성
candidate1 = test1.copy()
candidate1["qty"] = pred_candidate1

# 파일 구조 확인
assert len(candidate1) == len(test1)
assert candidate1.columns.tolist() == test1.columns.tolist() + ["qty"]
assert candidate1["qty"].notna().all()

# 후보 파일 저장
joblib.dump(xgb1_candidate, model_path)
candidate1.to_csv(csv_path, index=False)

print("전체 학습 데이터:", X_full.shape)
print("테스트 데이터:", X_test.shape)
print("학습 횟수:", params["n_estimators"])
print("모델 저장:", model_path.name)
print("CSV 저장:", csv_path.name)

print("\n예측 결과 상위 5행")
display(candidate1[["id", "Date", "product", "qty"]].head())
````

## 2. Text 문제 - 뉴스 제목 주제 분류(3종) (35점)

<sub>학습 목표: 뉴스 제목 주제 분류(3종) 모델로 테스트 정확도 0.91 이상을 달성할 수 있다 | 역량: PX1 텍스트 정제·벡터화, PX2 텍스트 분류 모델 학습·평가 | 기술: pandas, scikit-learn(CountVectorizer, TfidfVectorizer, MultinomialNB, LinearSVC), joblib | 난이도: 중급 | 예상 시간: 60분</sub>

### 문제 개요

뉴스 기사 **제목(평균 약 27자)** 만으로 기사의 **주제 3종** 을 분류하는 자연어 처리 문제입니다. 공식 샘플문항(가공식품 상품명 평균 19자 → 카테고리 8종)과 같은 '짧은 한국어 문장 분류' 형식입니다.

제공된 훈련 데이터셋을 활용해 텍스트 다중 분류(3종) 모델을 개발하고, 테스트 데이터셋의 'topic' 값을 예측하시기 바랍니다.

### 달성 목표

"02_test_x.csv" 파일의 테스트 데이터셋을 기준으로 **정확도(Accuracy) 91% 이상**을 달성하는 모델을 개발하시기 바랍니다.

단, 테스트 데이터의 정답(Label 값)은 별도로 제공되지 않습니다.

````python

import pandas as pd
from pathlib import Path

# 데이터 불러오기
base = Path("/content")

train2 = pd.read_csv(base / "02_train.csv")
test2 = pd.read_csv(base / "02_test_x.csv")

print("학습 데이터 크기:", train2.shape)
print("테스트 데이터 크기:", test2.shape)

print("\n학습 데이터 상위 5행")
display(train2.head())

print("\n테스트 데이터 상위 5행")
display(test2.head())

print("\n학습 데이터 정보")
train2.info()

print("\n주제별 데이터 개수")
print(train2["topic"].value_counts())

print("\n주제별 데이터 비율")
print(train2["topic"].value_counts(normalize=True).round(3))

print("\n뉴스 제목 길이")
print(train2["title"].str.len().describe())

print("\n결측치")
print("학습:", train2.isna().sum().to_dict())
print("테스트:", test2.isna().sum().to_dict())

print("\n중복된 뉴스 제목:", train2["title"].duplicated().sum())
````

````python

from sklearn.model_selection import train_test_split

# 입력 데이터와 정답 분리
X2 = train2["title"]
y2 = train2["topic"]

# 학습 80%, 검증 20% 분리
X2_train, X2_valid, y2_train, y2_valid = train_test_split(
    X2,
    y2,
    test_size=0.2,
    random_state=42,
    stratify=y2
)

print("학습 데이터:", X2_train.shape)
print("검증 데이터:", X2_valid.shape)

print("\n학습 주제별 개수")
print(y2_train.value_counts())

print("\n검증 주제별 개수")
print(y2_valid.value_counts())

print("\n학습·검증 데이터 중복 여부")
print("인덱스 중복:", len(set(X2_train.index) & set(X2_valid.index)))
````

````python

from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report

# 첫 번째 후보 모델
word_svc2 = make_pipeline(
    TfidfVectorizer(
        analyzer="word",
        ngram_range=(1, 2),
        min_df=2,
        sublinear_tf=True
    ),
    LinearSVC(C=1.0, random_state=42)
)

# 모델 학습
word_svc2.fit(X2_train, y2_train)

# 검증 데이터 예측
pred_word2 = word_svc2.predict(X2_valid)

# 정확도 평가
acc_word2 = accuracy_score(y2_valid, pred_word2)

print("모델: 단어 TF-IDF + LinearSVC")
print("검증 정확도:", round(acc_word2 * 100, 2), "%")
print("목표 정확도 91% 이상:", acc_word2 >= 0.91)

print("\n주제별 분류 성능")
print(classification_report(y2_valid, pred_word2))
````

````python

# 두 번째 후보 모델
char_svc2 = make_pipeline(
    TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(2, 5),
        min_df=2,
        sublinear_tf=True
    ),
    LinearSVC(C=1.0, random_state=42)
)

# 모델 학습
char_svc2.fit(X2_train, y2_train)

# 검증 데이터 예측
pred_char2 = char_svc2.predict(X2_valid)

# 정확도 평가
acc_char2 = accuracy_score(y2_valid, pred_char2)

print("모델: 문자 TF-IDF + LinearSVC")
print("검증 정확도:", round(acc_char2 * 100, 2), "%")
print("목표 정확도 91% 이상:", acc_char2 >= 0.91)

print("\n모델 성능 비교")
print("단어 TF-IDF:", round(acc_word2 * 100, 2), "%")
print("문자 TF-IDF:", round(acc_char2 * 100, 2), "%")

print("\n주제별 분류 성능")
print(classification_report(y2_valid, pred_char2))
````

````python

from sklearn.model_selection import StratifiedKFold, cross_val_score

# 5겹 교차검증 설정
cv2 = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# 단어 단위 모델 교차검증
word_scores2 = cross_val_score(
    word_svc2, X2_train, y2_train,
    cv=cv2, scoring="accuracy"
)

# 문자 단위 모델 교차검증
char_scores2 = cross_val_score(
    char_svc2, X2_train, y2_train,
    cv=cv2, scoring="accuracy"
)

print("단어 TF-IDF 교차검증")
print("각 Fold:", word_scores2.round(4))
print("평균 정확도:", round(word_scores2.mean() * 100, 2), "%")

print("\n문자 TF-IDF 교차검증")
print("각 Fold:", char_scores2.round(4))
print("평균 정확도:", round(char_scores2.mean() * 100, 2), "%")

print("\n기존 문자 모델 검증 정확도:",
      round(acc_char2 * 100, 2), "%")
````

````python

import joblib
from pathlib import Path
from sklearn.base import clone

model_path2 = Path("/content/candidate_2.pkl")
csv_path2 = Path("/content/candidate_2.csv")

# 기존 후보 파일 덮어쓰기 방지
if model_path2.exists() or csv_path2.exists():
    raise FileExistsError("기존 candidate_2 파일이 있습니다.")

# 기존 모델의 설정을 복사
char_svc2_candidate = clone(char_svc2)

# 전체 학습 데이터로 재학습
char_svc2_candidate.fit(train2["title"], train2["topic"])

# 테스트 데이터 예측
pred_candidate2 = char_svc2_candidate.predict(test2["title"])

# 후보 CSV 생성
candidate2 = test2.copy()
candidate2["topic"] = pred_candidate2

# 파일 구조 확인
assert len(candidate2) == len(test2)
assert candidate2.columns.tolist() == ["id", "title", "topic"]
assert candidate2["id"].equals(test2["id"])
assert candidate2["topic"].isin(train2["topic"].unique()).all()

# 후보 파일 저장
joblib.dump(char_svc2_candidate, model_path2)
candidate2.to_csv(csv_path2, index=False)

print("전체 학습 데이터:", train2.shape)
print("테스트 데이터:", test2.shape)
print("모델 저장:", model_path2.name)
print("CSV 저장:", csv_path2.name)

print("\n예측 결과 상위 5행")
display(candidate2.head())

print("\n예측된 주제별 개수")
print(candidate2["topic"].value_counts())
````

````python

import zipfile
import pandas as pd
from pathlib import Path
from PIL import Image

base = Path("/content")
zip_path = base / "03_images.zip"

# CSV 파일 경로 확인
def find_csv(names):
    for name in names:
        path = base / name
        if path.exists():
            return path
    raise FileNotFoundError(f"CSV 파일을 찾지 못했습니다: {names}")

train_path3 = find_csv(["03_train.csv", "03_train(1).csv"])
test_path3 = find_csv(["03_test_x.csv", "03_test_x(1).csv"])

train3 = pd.read_csv(train_path3)
test3 = pd.read_csv(test_path3)

# 기존 압축 해제 폴더가 있으면 재사용
if (base / "03_train").is_dir() and (base / "03_test").is_dir():
    IMG_DIR = base
else:
    IMG_DIR = base / "aice_05_images"

train_dir3 = IMG_DIR / "03_train"
test_dir3 = IMG_DIR / "03_test"

# 필요한 경우에만 압축 해제
if not train_dir3.exists() or not test_dir3.exists():
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(IMG_DIR)

print("학습 데이터:", train3.shape)
print("테스트 데이터:", test3.shape)

print("\n학습 데이터 예시")
display(train3.head())

print("\n테스트 데이터 예시")
display(test3.head())

print("\n라벨별 이미지 개수")
print(train3["label"].value_counts().sort_index())

print("\n이미지 파일 개수")
print("학습:", len(list(train_dir3.rglob("*.png"))))
print("테스트:", len(list(test_dir3.glob("*.png"))))

# 실제 이미지 확인
sample = train3.iloc[0]
sample_path = train_dir3 / sample["label"] / sample["file_name"]

with Image.open(sample_path) as img:
    print("\n이미지 크기:", img.size)
    print("이미지 모드:", img.mode)

# CSV에 적힌 이미지가 실제 존재하는지 확인
train_ok = all(
    (train_dir3 / row.label / row.file_name).is_file()
    for row in train3.itertuples()
)

test_ok = all(
    (test_dir3 / name).is_file()
    for name in test3["file_name"]
)

print("\n학습 이미지 경로 정상:", train_ok)
print("테스트 이미지 경로 정상:", test_ok)
````

````python

from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 이미지 정규화 및 검증 데이터 비율 설정
gen3 = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

# 이미지 불러오기 공통 설정
img_options3 = dict(
    target_size=(28, 28),
    color_mode="grayscale",
    class_mode="categorical",
    batch_size=32,
    seed=42
)

# 학습 이미지
train_gen3 = gen3.flow_from_directory(
    str(train_dir3),
    subset="training",
    shuffle=True,
    **img_options3
)

# 검증 이미지
valid_gen3 = gen3.flow_from_directory(
    str(train_dir3),
    subset="validation",
    shuffle=False,
    **img_options3
)

print("\n학습 이미지:", train_gen3.samples)
print("검증 이미지:", valid_gen3.samples)

print("\n클래스 번호")
print(train_gen3.class_indices)

# 이미지 데이터 형태 확인
images3, labels3 = next(train_gen3)

print("\n이미지 배치 크기:", images3.shape)
print("정답 배치 크기:", labels3.shape)
print("픽셀 최솟값:", images3.min())
print("픽셀 최댓값:", images3.max())

print("\n클래스 일치:",
      train_gen3.class_indices == valid_gen3.class_indices)

# 다음 학습을 위해 시작 위치 초기화
train_gen3.reset()
````

````python

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D,
    Flatten, Dense, Dropout
)
from tensorflow.keras.optimizers import Adam

tf.keras.utils.set_random_seed(42)

# CNN 후보 모델 생성
cnn3 = Sequential([
    Input(shape=(28, 28, 1)),

    Conv2D(32, (3, 3), activation="relu", padding="same"),
    Conv2D(32, (3, 3), activation="relu", padding="same"),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation="relu", padding="same"),
    Conv2D(64, (3, 3), activation="relu", padding="same"),
    MaxPooling2D((2, 2)),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.4),
    Dense(10, activation="softmax")
])

# 모델 학습 설정
cnn3.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("CNN 모델 구조")
cnn3.summary()

print("\n입력 형태:", cnn3.input_shape)
print("출력 형태:", cnn3.output_shape)
print("학습 가능 파라미터:", cnn3.count_params())
````

````python

from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# 조기 종료
early3 = EarlyStopping(
    monitor="val_loss",
    patience=6,
    restore_best_weights=True,
    verbose=1
)

# 학습률 조정
reduce3 = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    min_lr=0.00001,
    verbose=1
)

# 이미지 데이터 시작 위치 초기화
train_gen3.reset()
valid_gen3.reset()

# CNN 모델 학습
history3 = cnn3.fit(
    train_gen3,
    validation_data=valid_gen3,
    epochs=35,
    callbacks=[early3, reduce3],
    verbose=2
)

# 학습 완료 후 실제 성능 평가
train_loss3, train_acc3 = cnn3.evaluate(
    train_gen3, verbose=0
)

valid_loss3, valid_acc3 = cnn3.evaluate(
    valid_gen3, verbose=0
)

print("\nCNN 최종 평가 결과")
print("학습 정확도:", round(train_acc3 * 100, 2), "%")
print("검증 정확도:", round(valid_acc3 * 100, 2), "%")
print("검증 손실:", round(valid_loss3, 4))
print("실제 학습 횟수:", len(history3.history["loss"]))

print("\n목표 정확도 80% 이상:", valid_acc3 >= 0.80)
print("자동 채점 기준 72% 이상:", valid_acc3 >= 0.72)
````

````python

import numpy as np
from pathlib import Path
from tensorflow.keras.models import load_model

model_path3 = Path("/content/candidate_3.h5")

# 기존 후보 파일 덮어쓰기 방지
if model_path3.exists():
    raise FileExistsError("candidate_3.h5 파일이 이미 존재합니다.")

# 학습한 CNN 모델 저장
cnn3.save(str(model_path3))

# 저장된 모델 다시 불러오기
loaded_cnn3 = load_model(str(model_path3), compile=False)

# 저장 전 모델 예측
valid_gen3.reset()
pred_before3 = cnn3.predict(valid_gen3, verbose=0)

# 저장 후 모델 예측
valid_gen3.reset()
pred_after3 = loaded_cnn3.predict(valid_gen3, verbose=0)

# 예측 클래스 번호 추출
pred_classes3 = np.argmax(pred_after3, axis=1)

# 실제 정답 클래스 번호
true_classes3 = valid_gen3.classes

# 예측 정확도 계산
predict_acc3 = np.mean(pred_classes3 == true_classes3)

print("후보 모델 저장:", model_path3.name)
print("모델 파일 크기:", round(model_path3.stat().st_size / 1024**2, 2), "MB")

print("\n저장 전후 예측 일치:",
      np.allclose(pred_before3, pred_after3, atol=1e-5))

print("검증 이미지 개수:", len(pred_classes3))
print("predict 정확도:", round(predict_acc3 * 100, 2), "%")
print("evaluate 정확도:", round(valid_acc3 * 100, 2), "%")

print("\n예측 결과와 평가 일치:",
      np.isclose(predict_acc3, valid_acc3, atol=1e-4))
print("목표 정확도 80% 이상:", predict_acc3 >= 0.80)
````

````python

import numpy as np
import pandas as pd
from pathlib import Path
from tensorflow.keras.preprocessing.image import ImageDataGenerator

csv_path3 = Path("/content/candidate_3.csv")

if csv_path3.exists():
    raise FileExistsError("candidate_3.csv 파일이 이미 존재합니다.")

# 테스트 이미지 불러오기
test_img_gen3 = ImageDataGenerator(rescale=1./255)

test_gen3 = test_img_gen3.flow_from_directory(
    str(IMG_DIR),
    classes=["03_test"],
    target_size=(28, 28),
    color_mode="grayscale",
    class_mode=None,
    batch_size=32,
    shuffle=False
)

# 저장된 CNN 모델로 예측
test_gen3.reset()
pred_probs3 = loaded_cnn3.predict(test_gen3, verbose=1)

# 예측 확률에서 클래스 번호 추출
pred_nums3 = np.argmax(pred_probs3, axis=1)

# 클래스 번호를 실제 라벨 이름으로 변환
idx_to_label3 = {
    v: k for k, v in train_gen3.class_indices.items()
}

pred_labels3 = [
    idx_to_label3[i] for i in pred_nums3
]

# 이미지 파일명과 예측값 연결
test_names3 = [
    Path(name).name for name in test_gen3.filenames
]

assert len(set(test_names3)) == len(test_names3)
assert np.isfinite(pred_probs3).all()

pred_map3 = pd.Series(pred_labels3, index=test_names3)

# 원본 테스트 CSV 순서에 맞춰 예측값 추가
candidate3 = test3.copy()
candidate3["label"] = candidate3["file_name"].map(pred_map3)

# 제출 형식 확인
assert len(candidate3) == len(test3)
assert candidate3.columns.tolist() == ["file_name", "label"]
assert candidate3["file_name"].equals(test3["file_name"])
assert candidate3["label"].notna().all()
assert candidate3["label"].isin(train3["label"].unique()).all()

# 후보 CSV 저장
candidate3.to_csv(csv_path3, index=False)

print("\n테스트 이미지:", test_gen3.samples)
print("예측 결과:", len(pred_labels3))
print("후보 파일 저장:", csv_path3.name)

print("\n예측 결과 상위 5행")
display(candidate3.head())

print("\n클래스별 예측 개수")
print(candidate3["label"].value_counts())

print("\n테스트 행 순서 유지:",
      candidate3["file_name"].equals(test3["file_name"]))

print("결측치:", candidate3.isna().sum().sum())
````
